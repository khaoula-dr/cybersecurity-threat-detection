```python

import pandas as pd, numpy as np
import matplotlib.pyplot as plt, seaborn as sns

train = pd.read_csv("../data/raw/UNSW_NB15_training-set.csv")
test  = pd.read_csv("../data/raw/UNSW_NB15_testing-set.csv")
print(train.shape, test.shape)
train.info()
train.head()
```

```python

for name, df in [("train", train), ("test", test)]:
    print(name, df["label"].value_counts(normalize=True).round(3).to_dict())

print(train["attack_cat"].value_counts())
print("scale_pos_weight =", (train.label == 0).sum() / (train.label == 1).sum())
```

```python

print("Doublons :", train.duplicated().sum())
print("Manquants :\n", train.isna().sum()[train.isna().sum() > 0])
cat_cols = train.select_dtypes(include="object").columns.tolist()
num_cols = train.select_dtypes(exclude="object").columns.drop(["label"]).tolist()
print("Catégorielles :", cat_cols)
print(train[cat_cols].nunique())
```

```python
#statistiques et valeurs extrêmes
train[num_cols].describe().T[["mean", "50%", "max"]]
```

```python
#corrélations avec la cible
corr = train[num_cols + ["label"]].corr()["label"].drop("label").sort_values()
print(corr.head(8)); print(corr.tail(8))
plt.figure(figsize=(10, 8))
sns.heatmap(train[num_cols].corr(), cmap="coolwarm", center=0)
```

```python
#différence train / test officiels
for c in ["proto", "service", "state"]:
    unseen = set(test[c]) - set(train[c])
    print(c, "valeurs absentes du train :", unseen)
```
